The text format necessitates first executing the following Python chunk:

````python
import re, sqlite3

with open("Movies-Database/moviedb.sql", encoding="utf-8") as f:
    sql = f.read()

sql = re.sub(r"/\*!.*?\*/;", "", sql)
sql = re.sub(r"^SET .*?;$", "", sql, flags=re.M)
sql = re.sub(r"\) ENGINE=.*?;", ");", sql)
sql = sql.replace("AUTO_INCREMENT", "")
sql = sql.replace("START TRANSACTION;", "BEGIN;")
sql = sql.replace("\\'", "''").replace('\\"', '"').replace("\\r", "\r")

con = sqlite3.connect(":memory:")
con.executescript(sql); # now con is ready for normal operation
````
